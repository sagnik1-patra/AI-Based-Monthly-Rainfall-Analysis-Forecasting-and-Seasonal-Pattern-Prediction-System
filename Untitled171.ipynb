```python
from pathlib import Path
import json
import random
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import yaml
import joblib

from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads\AI-Based Monthly Rainfall Analysis, Forecasting, and Seasonal Pattern Prediction System"
)

CSV_PATH = ROOT / "D67. Rainfall Monthly Total_1_0.csv"
ROOT.mkdir(parents=True, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

CONFIG = {
    "project": "AI-Based Monthly Rainfall Forecasting",
    "dataset": str(CSV_PATH),
    "target": "Total_rainfall",
    "date_column": "Month-Year",
    "random_seed": SEED,
    "test_size": 0.20,
    "forecast_months": 12,
    "lags": [1, 2, 3, 6, 12],
    "rolling_windows": [3, 6],
    "neural_network": {
        "epochs": 200,
        "batch_size": 8,
        "learning_rate": 0.001
    }
}

with open(ROOT / "config.yaml", "w", encoding="utf-8") as f:
    yaml.safe_dump(CONFIG, f, sort_keys=False)

print("Configuration saved.")

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

required = ["Month-Year", "Total_rainfall"]
missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(f"Missing columns: {missing}")

df["Month-Year"] = pd.to_datetime(
    df["Month-Year"],
    format="%b-%y",
    errors="coerce"
)

df["Total_rainfall"] = pd.to_numeric(
    df["Total_rainfall"],
    errors="coerce"
)

df = df.dropna(subset=required)
df = df.sort_values("Month-Year").reset_index(drop=True)

if df["Month-Year"].duplicated().any():
    raise ValueError("Duplicate months found in dataset.")

expected_dates = pd.date_range(
    df["Month-Year"].min(),
    df["Month-Year"].max(),
    freq="MS"
)

if len(expected_dates) != len(df) or not (
    expected_dates == pd.DatetimeIndex(df["Month-Year"])
).all():
    raise ValueError(
        "Dataset must contain continuous monthly observations."
    )

if (df["Total_rainfall"] < 0).any():
    raise ValueError("Rainfall cannot be negative.")

print("\nDataset loaded successfully!")
print(df.head())
print("\nDataset shape:", df.shape)
print("\nDate range:", df["Month-Year"].min(),
      "to", df["Month-Year"].max())

# ============================================================
# 3. FEATURE ENGINEERING
# ============================================================

LAGS = CONFIG["lags"]
WINDOWS = CONFIG["rolling_windows"]

FEATURE_COLUMNS = [
    "month",
    "month_sin",
    "month_cos",
    "time_index"
]

FEATURE_COLUMNS += [f"lag_{lag}" for lag in LAGS]
FEATURE_COLUMNS += [
    f"rolling_mean_{w}" for w in WINDOWS
]
FEATURE_COLUMNS += [
    f"rolling_std_{w}" for w in WINDOWS
]

def create_features(data):
    result = data.copy()

    result["month"] = result["Month-Year"].dt.month
    result["month_sin"] = np.sin(
        2 * np.pi * result["month"] / 12
    )
    result["month_cos"] = np.cos(
        2 * np.pi * result["month"] / 12
    )
    result["time_index"] = np.arange(len(result))

    for lag in LAGS:
        result[f"lag_{lag}"] = (
            result["Total_rainfall"].shift(lag)
        )

    previous = result["Total_rainfall"].shift(1)

    for window in WINDOWS:
        result[f"rolling_mean_{window}"] = (
            previous.rolling(window).mean()
        )
        result[f"rolling_std_{window}"] = (
            previous.rolling(window).std()
        )

    return result

featured = create_features(df)
featured = featured.dropna().reset_index(drop=True)

X = featured[FEATURE_COLUMNS]
y = featured["Total_rainfall"]

if len(featured) < 30:
    raise ValueError("Insufficient data after feature engineering.")

print("\nFeature engineering completed.")
print("Features:", FEATURE_COLUMNS)
print("Usable records:", len(featured))

# ============================================================
# 4. CHRONOLOGICAL TRAIN / TEST SPLIT
# ============================================================

split_index = int(len(X) * (1 - CONFIG["test_size"]))

X_train = X.iloc[:split_index].copy()
X_test = X.iloc[split_index:].copy()

y_train = y.iloc[:split_index].copy()
y_test = y.iloc[split_index:].copy()

test_dates = featured["Month-Year"].iloc[split_index:]

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# 5. TRAIN SCIKIT-LEARN MODELS
# ============================================================

models = {
    "Random Forest": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", RandomForestRegressor(
            n_estimators=200,
            max_depth=8,
            min_samples_leaf=2,
            random_state=SEED,
            n_jobs=-1
        ))
    ]),

    "Gradient Boosting": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", GradientBoostingRegressor(
            n_estimators=150,
            learning_rate=0.05,
            max_depth=2,
            random_state=SEED
        ))
    ])
}

def calculate_metrics(actual, predicted):
    return {
        "MAE": float(mean_absolute_error(actual, predicted)),
        "RMSE": float(np.sqrt(
            mean_squared_error(actual, predicted)
        )),
        "R2": float(r2_score(actual, predicted))
    }

results = {}
predictions = {}

for name, model in models.items():
    print(f"\nTraining {name}...")

    model.fit(X_train, y_train)

    pred = np.maximum(model.predict(X_test), 0)

    results[name] = calculate_metrics(y_test, pred)
    predictions[name] = pred

    print(results[name])

# ============================================================
# 6. TRAIN KERAS NEURAL NETWORK
# ============================================================

print("\nTraining Neural Network...")

imputer_nn = SimpleImputer(strategy="median")
scaler_nn = StandardScaler()

X_train_nn = imputer_nn.fit_transform(X_train)
X_train_nn = scaler_nn.fit_transform(X_train_nn)

X_test_nn = imputer_nn.transform(X_test)
X_test_nn = scaler_nn.transform(X_test_nn)

# Validation is taken from the end of the training period.
validation_count = max(8, int(len(X_train_nn) * 0.2))

X_fit = X_train_nn[:-validation_count]
X_val = X_train_nn[-validation_count:]

y_fit = y_train.to_numpy()[:-validation_count]
y_val = y_train.to_numpy()[-validation_count:]

nn = Sequential([
    Input(shape=(X_train_nn.shape[1],)),
    Dense(64, activation="relu"),
    Dropout(0.15),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(1, activation="linear")
])

nn.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

history = nn.fit(
    X_fit,
    y_fit,
    validation_data=(X_val, y_val),
    epochs=CONFIG["neural_network"]["epochs"],
    batch_size=CONFIG["neural_network"]["batch_size"],
    callbacks=[early_stop],
    verbose=0,
    shuffle=False
)

nn_pred = nn.predict(X_test_nn, verbose=0).flatten()
nn_pred = np.maximum(nn_pred, 0)

results["Neural Network"] = calculate_metrics(
    y_test, nn_pred
)
predictions["Neural Network"] = nn_pred

print("Neural Network:", results["Neural Network"])

# ============================================================
# 7. SELECT BEST MODEL
# ============================================================

best_model_name = min(
    results,
    key=lambda name: results[name]["RMSE"]
)

print("\nBest model:", best_model_name)

# ============================================================
# 8. REFIT FINAL MODELS ON ALL AVAILABLE DATA
# ============================================================

# Test performance above is held out.
# Final models below are refitted for future forecasting.

for name, model in models.items():
    model.fit(X, y)

final_imputer = SimpleImputer(strategy="median")
final_scaler = StandardScaler()

X_all_nn = final_imputer.fit_transform(X)
X_all_nn = final_scaler.fit_transform(X_all_nn)

# Refit the neural network on all available samples.
tf.keras.backend.clear_session()
tf.random.set_seed(SEED)

final_nn = Sequential([
    Input(shape=(X_all_nn.shape[1],)),
    Dense(64, activation="relu"),
    Dropout(0.15),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(1, activation="linear")
])

final_nn.compile(
    optimizer=tf.keras.optimizers.Adam(0.001),
    loss="mse",
    metrics=["mae"]
)

final_nn.fit(
    X_all_nn,
    y.to_numpy(),
    epochs=max(1, len(history.history["loss"])),
    batch_size=8,
    verbose=0,
    shuffle=False
)

# ============================================================
# 9. SAVE H5 MODEL
# ============================================================

H5_PATH = ROOT / "rainfall_model.h5"

final_nn.save(str(H5_PATH))

print("\nH5 model saved:", H5_PATH)

# ============================================================
# 10. SAVE PKL MODEL
# ============================================================

# The PKL file contains the selected classical model.
# If NN wins, it also contains the best classical alternative.

best_classical = min(
    models,
    key=lambda name: results[name]["RMSE"]
)

pkl_bundle = {
    "model": models[best_classical],
    "model_name": best_classical,
    "feature_columns": FEATURE_COLUMNS,
    "lags": LAGS,
    "rolling_windows": WINDOWS,
    "neural_network_imputer": final_imputer,
    "neural_network_scaler": final_scaler,
    "selected_best_model": best_model_name,
    "last_observed_date": str(
        df["Month-Year"].iloc[-1].date()
    )
}

PKL_PATH = ROOT / "rainfall_model.pkl"

joblib.dump(pkl_bundle, PKL_PATH)

print("PKL model saved:", PKL_PATH)

# ============================================================
# 11. SAVE TEST RESULTS
# ============================================================

test_results = pd.DataFrame({
    "Month-Year": test_dates.to_numpy(),
    "Actual_Rainfall": y_test.to_numpy()
})

for name, pred in predictions.items():
    column = name.replace(" ", "_") + "_Prediction"
    test_results[column] = pred

test_results.to_csv(
    ROOT / "result.csv",
    index=False
)

# ============================================================
# 12. FUTURE RAINFALL FORECASTING
# ============================================================

def build_future_features(history_df, next_date):
    temp = pd.concat([
        history_df,
        pd.DataFrame({
            "Month-Year": [next_date],
            "Total_rainfall": [np.nan]
        })
    ], ignore_index=True)

    # Feature creation uses only previous observations.
    temp_features = create_features(temp)

    return temp_features[FEATURE_COLUMNS].iloc[[-1]]

def predict_future(model_name, future_X):
    if model_name == "Neural Network":
        transformed = final_imputer.transform(future_X)
        transformed = final_scaler.transform(transformed)

        value = final_nn.predict(
            transformed, verbose=0
        ).flatten()[0]
    else:
        value = models[model_name].predict(future_X)[0]

    return max(float(value), 0.0)

forecast_history = df.copy()
future_records = []

for step in range(CONFIG["forecast_months"]):
    next_date = (
        forecast_history["Month-Year"].iloc[-1]
        + pd.DateOffset(months=1)
    )

    future_X = build_future_features(
        forecast_history, next_date
    )

    future_value = predict_future(
        best_model_name, future_X
    )

    future_records.append({
        "Month-Year": next_date.strftime("%Y-%m"),
        "Predicted_Rainfall": round(future_value, 3)
    })

    forecast_history = pd.concat([
        forecast_history,
        pd.DataFrame({
            "Month-Year": [next_date],
            "Total_rainfall": [future_value]
        })
    ], ignore_index=True)

forecast_df = pd.DataFrame(future_records)

forecast_df.to_csv(
    ROOT / "prediction.csv",
    index=False
)

print("\nFuture rainfall predictions:")
print(forecast_df.to_string(index=False))

# ============================================================
# 13. SAVE MODEL INFORMATION AS JSON
# ============================================================

model_info = {
    "project_name": CONFIG["project"],
    "created_at": datetime.now().isoformat(),
    "dataset": CSV_PATH.name,
    "total_records": int(len(df)),
    "training_samples": int(len(X_train)),
    "testing_samples": int(len(X_test)),
    "feature_columns": FEATURE_COLUMNS,
    "target_column": "Total_rainfall",
    "forecast_horizon_months": CONFIG["forecast_months"],
    "best_model": best_model_name,
    "best_classical_model": best_classical,
    "selection_metric": "RMSE",
    "metrics": results,
    "saved_files": {
        "h5": H5_PATH.name,
        "pkl": PKL_PATH.name,
        "yaml": "config.yaml",
        "json": "model_info.json",
        "test_results": "result.csv",
        "future_predictions": "prediction.csv"
    },
    "notes": [
        "Chronological train-test split used.",
        "Lag features use past rainfall only.",
        "H5 stores the neural network.",
        "PKL stores the best classical model.",
        "Future predictions are recursive estimates.",
        "Reported metrics are held-out test metrics."
    ]
}

JSON_PATH = ROOT / "model_info.json"

with open(JSON_PATH, "w", encoding="utf-8") as f:
    json.dump(model_info, f, indent=4)

print("\nJSON metadata saved:", JSON_PATH)

# ============================================================
# 14. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("RAINFALL FORECASTING PROJECT COMPLETED")
print("=" * 60)

for name, metric in results.items():
    print(
        f"{name}: "
        f"MAE={metric['MAE']:.3f}, "
        f"RMSE={metric['RMSE']:.3f}, "
        f"R2={metric['R2']:.3f}"
    )

print("\nSelected model:", best_model_name)
print("\nSaved files:")

for filename in [
    "rainfall_model.h5",
    "rainfall_model.pkl",
    "config.yaml",
    "model_info.json",
    "result.csv",
    "prediction.csv"
]:
    print(" -", ROOT / filename)

print("\nAll operations completed successfully!")
```
