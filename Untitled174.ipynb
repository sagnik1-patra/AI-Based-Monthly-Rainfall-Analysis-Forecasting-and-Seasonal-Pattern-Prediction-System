```python
from pathlib import Path
from datetime import datetime
import json
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
import joblib
import tensorflow as tf

from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ==============================================================
# 1. PROJECT CONFIGURATION
# ==============================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads\AI-Based Monthly Rainfall Analysis, Forecasting, and Seasonal Pattern Prediction System"
)

ROOT.mkdir(parents=True, exist_ok=True)

CSV_PATH = ROOT / "D67. Rainfall Monthly Total_1_0.csv"

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

rng = np.random.default_rng(SEED)

sns.set_theme(style="whitegrid")

CONFIG = {
    "project": "PSO-Based Monthly Rainfall Forecasting",
    "algorithm": "Particle Swarm Optimization",
    "dataset": str(CSV_PATH),
    "seed": SEED,
    "forecast_months": 12,
    "lags": [1, 2, 3, 6, 12],
    "rolling_windows": [3, 6],
    "train_ratio": 0.60,
    "validation_ratio": 0.20,
    "test_ratio": 0.20,
    "pso_random_forest": {
        "particles": 10,
        "iterations": 12,
        "inertia": 0.7,
        "cognitive": 1.5,
        "social": 1.5
    },
    "pso_neural_network": {
        "particles": 5,
        "iterations": 5,
        "search_epochs": 50,
        "final_epochs": 80
    }
}

with open(
    ROOT / "pso_config.yaml",
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(CONFIG, f, sort_keys=False)

print("=" * 65)
print("PSO-BASED MONTHLY RAINFALL FORECASTING")
print("=" * 65)

# ==============================================================
# 2. LOAD DATASET
# ==============================================================

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

required_columns = [
    "Month-Year",
    "Total_rainfall"
]

for column in required_columns:
    if column not in df.columns:
        raise ValueError(
            f"Missing required column: {column}"
        )

df["Month-Year"] = pd.to_datetime(
    df["Month-Year"],
    format="%b-%y",
    errors="coerce"
)

df["Total_rainfall"] = pd.to_numeric(
    df["Total_rainfall"],
    errors="coerce"
)

df = df.dropna(
    subset=required_columns
)

df = df.sort_values(
    "Month-Year"
).reset_index(drop=True)

if df["Month-Year"].duplicated().any():
    raise ValueError("Duplicate months detected.")

if (df["Total_rainfall"] < 0).any():
    raise ValueError("Negative rainfall detected.")

expected_dates = pd.date_range(
    df["Month-Year"].min(),
    df["Month-Year"].max(),
    freq="MS"
)

if not np.array_equal(
    expected_dates.to_numpy(),
    df["Month-Year"].to_numpy()
):
    raise ValueError(
        "Dataset must contain continuous monthly records."
    )

print("\nDataset loaded successfully!")
print(df.head())
print("\nDataset shape:", df.shape)

# ==============================================================
# 3. FEATURE ENGINEERING
# ==============================================================

LAGS = CONFIG["lags"]
WINDOWS = CONFIG["rolling_windows"]

FEATURE_COLUMNS = [
    "month",
    "month_sin",
    "month_cos",
    "time_index"
]

FEATURE_COLUMNS += [
    f"lag_{lag}" for lag in LAGS
]

FEATURE_COLUMNS += [
    f"rolling_mean_{w}" for w in WINDOWS
]

FEATURE_COLUMNS += [
    f"rolling_std_{w}" for w in WINDOWS
]

def create_features(data):

    result = data.copy()

    result["month"] = result["Month-Year"].dt.month

    result["month_sin"] = np.sin(
        2 * np.pi * result["month"] / 12
    )

    result["month_cos"] = np.cos(
        2 * np.pi * result["month"] / 12
    )

    result["time_index"] = np.arange(
        len(result)
    )

    for lag in LAGS:

        result[f"lag_{lag}"] = (
            result["Total_rainfall"].shift(lag)
        )

    previous = result["Total_rainfall"].shift(1)

    for window in WINDOWS:

        result[f"rolling_mean_{window}"] = (
            previous.rolling(window).mean()
        )

        result[f"rolling_std_{window}"] = (
            previous.rolling(window).std()
        )

    return result

featured = create_features(df)

featured = featured.dropna().reset_index(drop=True)

X = featured[FEATURE_COLUMNS]
y = featured["Total_rainfall"]
dates = featured["Month-Year"]

print("\nFeature engineering completed")
print("Feature count:", len(FEATURE_COLUMNS))
print("Usable samples:", len(X))

# ==============================================================
# 4. CHRONOLOGICAL DATA SPLIT
# ==============================================================

n = len(X)

train_end = int(n * 0.60)
validation_end = int(n * 0.80)

X_train = X.iloc[:train_end]
y_train = y.iloc[:train_end]

X_val = X.iloc[train_end:validation_end]
y_val = y.iloc[train_end:validation_end]

X_test = X.iloc[validation_end:]
y_test = y.iloc[validation_end:]

test_dates = dates.iloc[validation_end:]

X_train_val = X.iloc[:validation_end]
y_train_val = y.iloc[:validation_end]

print("\nTraining samples:", len(X_train))
print("Validation samples:", len(X_val))
print("Testing samples:", len(X_test))

# ==============================================================
# 5. METRIC FUNCTIONS
# ==============================================================

def evaluate(actual, predicted):

    predicted = np.maximum(
        np.asarray(predicted).reshape(-1),
        0
    )

    return {
        "MAE": float(
            mean_absolute_error(actual, predicted)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(actual, predicted)
            )
        ),
        "R2": float(
            r2_score(actual, predicted)
        )
    }

def rmse(actual, predicted):

    return float(
        np.sqrt(
            mean_squared_error(actual, predicted)
        )
    )

# ==============================================================
# 6. BASELINE RANDOM FOREST
# ==============================================================

def create_rf(params):

    return Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "model",
            RandomForestRegressor(
                **params,
                random_state=SEED,
                n_jobs=-1
            )
        )
    ])

baseline_params = {
    "n_estimators": 100,
    "max_depth": 8,
    "min_samples_split": 2,
    "min_samples_leaf": 2
}

baseline = create_rf(baseline_params)

baseline.fit(
    X_train_val,
    y_train_val
)

baseline_pred = np.maximum(
    baseline.predict(X_test),
    0
)

baseline_metrics = evaluate(
    y_test,
    baseline_pred
)

print("\nBaseline Random Forest:")
print(baseline_metrics)

# ==============================================================
# 7. RANDOM FOREST PSO SEARCH SPACE
# ==============================================================

RF_BOUNDS = np.array([
    [50, 250],   # n_estimators
    [2, 15],     # max_depth
    [2, 12],     # min_samples_split
    [1, 6]       # min_samples_leaf
], dtype=float)

def decode_rf(position):

    values = np.rint(position).astype(int)

    return {
        "n_estimators": int(values[0]),
        "max_depth": int(values[1]),
        "min_samples_split": int(values[2]),
        "min_samples_leaf": int(values[3])
    }

rf_cache = {}

def rf_fitness(position):

    params = decode_rf(position)

    key = tuple(params.values())

    if key in rf_cache:
        return rf_cache[key]

    model = create_rf(params)

    model.fit(
        X_train,
        y_train
    )

    prediction = np.maximum(
        model.predict(X_val),
        0
    )

    score = rmse(
        y_val,
        prediction
    )

    rf_cache[key] = score

    return score

# ==============================================================
# 8. GENERIC PARTICLE SWARM OPTIMIZATION
# ==============================================================

def particle_swarm_optimization(
    fitness_function,
    bounds,
    n_particles=10,
    iterations=12,
    inertia=0.7,
    cognitive=1.5,
    social=1.5,
    label="PSO"
):

    dimensions = len(bounds)

    lower = bounds[:, 0]
    upper = bounds[:, 1]

    span = upper - lower

    positions = rng.uniform(
        lower,
        upper,
        size=(n_particles, dimensions)
    )

    velocities = rng.uniform(
        -0.1 * span,
        0.1 * span,
        size=(n_particles, dimensions)
    )

    personal_best_positions = positions.copy()

    personal_best_scores = np.array([
        fitness_function(position)
        for position in positions
    ])

    best_index = np.argmin(
        personal_best_scores
    )

    global_best_position = (
        personal_best_positions[best_index].copy()
    )

    global_best_score = float(
        personal_best_scores[best_index]
    )

    history = [global_best_score]

    for iteration in range(iterations):

        for i in range(n_particles):

            r1 = rng.random(dimensions)
            r2 = rng.random(dimensions)

            velocities[i] = (
                inertia * velocities[i]
                + cognitive * r1 * (
                    personal_best_positions[i]
                    - positions[i]
                )
                + social * r2 * (
                    global_best_position
                    - positions[i]
                )
            )

            velocities[i] = np.clip(
                velocities[i],
                -0.2 * span,
                0.2 * span
            )

            positions[i] = np.clip(
                positions[i] + velocities[i],
                lower,
                upper
            )

            score = fitness_function(
                positions[i]
            )

            if score < personal_best_scores[i]:

                personal_best_scores[i] = score

                personal_best_positions[i] = (
                    positions[i].copy()
                )

                if score < global_best_score:

                    global_best_score = score

                    global_best_position = (
                        positions[i].copy()
                    )

        history.append(global_best_score)

        print(
            f"{label} Iteration "
            f"{iteration + 1}/{iterations} | "
            f"Best Validation RMSE: "
            f"{global_best_score:.4f}"
        )

    return (
        global_best_position,
        global_best_score,
        history
    )

# ==============================================================
# 9. OPTIMIZE RANDOM FOREST USING PSO
# ==============================================================

print("\nStarting PSO Random Forest Optimization...")

rf_settings = CONFIG["pso_random_forest"]

best_rf_position, best_rf_score, rf_history = (
    particle_swarm_optimization(
        fitness_function=rf_fitness,
        bounds=RF_BOUNDS,
        n_particles=rf_settings["particles"],
        iterations=rf_settings["iterations"],
        inertia=rf_settings["inertia"],
        cognitive=rf_settings["cognitive"],
        social=rf_settings["social"],
        label="RF-PSO"
    )
)

best_rf_params = decode_rf(
    best_rf_position
)

print("\nBest RF Parameters:")
print(best_rf_params)

print("Best Validation RMSE:", best_rf_score)

# ==============================================================
# 10. TRAIN PSO-OPTIMIZED RANDOM FOREST
# ==============================================================

pso_rf = create_rf(
    best_rf_params
)

pso_rf.fit(
    X_train_val,
    y_train_val
)

pso_rf_pred = np.maximum(
    pso_rf.predict(X_test),
    0
)

pso_rf_metrics = evaluate(
    y_test,
    pso_rf_pred
)

print("\nPSO Random Forest Performance:")
print(pso_rf_metrics)

# ==============================================================
# 11. PREPARE NEURAL NETWORK DATA
# ==============================================================

nn_imputer = SimpleImputer(
    strategy="median"
)

nn_scaler = StandardScaler()

X_train_nn = nn_imputer.fit_transform(
    X_train
)

X_train_nn = nn_scaler.fit_transform(
    X_train_nn
)

X_val_nn = nn_scaler.transform(
    nn_imputer.transform(X_val)
)

# ==============================================================
# 12. NEURAL NETWORK SEARCH SPACE
# ==============================================================

# PSO searches:
# 1. Hidden layer neurons
# 2. Log10 learning rate

NN_BOUNDS = np.array([
    [16, 128],
    [-4, -2]
], dtype=float)

def decode_nn(position):

    units = int(
        np.rint(position[0])
    )

    learning_rate = float(
        10 ** position[1]
    )

    return units, learning_rate

def build_nn(units, learning_rate):

    model = tf.keras.Sequential([
        tf.keras.layers.Input(
            shape=(len(FEATURE_COLUMNS),)
        ),
        tf.keras.layers.Dense(
            units,
            activation="relu"
        ),
        tf.keras.layers.Dense(
            max(8, units // 2),
            activation="relu"
        ),
        tf.keras.layers.Dense(1)
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        ),
        loss="mse",
        metrics=["mae"]
    )

    return model

nn_cache = {}

def nn_fitness(position):

    units, learning_rate = decode_nn(
        position
    )

    key = (
        units,
        round(learning_rate, 7)
    )

    if key in nn_cache:
        return nn_cache[key]

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(SEED)

    model = build_nn(
        units,
        learning_rate
    )

    early_stop = tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    )

    model.fit(
        X_train_nn,
        y_train.to_numpy(),
        validation_data=(
            X_val_nn,
            y_val.to_numpy()
        ),
        epochs=CONFIG[
            "pso_neural_network"
        ]["search_epochs"],
        batch_size=8,
        callbacks=[early_stop],
        shuffle=False,
        verbose=0
    )

    prediction = np.maximum(
        model.predict(
            X_val_nn,
            verbose=0
        ).flatten(),
        0
    )

    score = rmse(
        y_val,
        prediction
    )

    nn_cache[key] = score

    return score

# ==============================================================
# 13. OPTIMIZE NEURAL NETWORK USING PSO
# ==============================================================

print("\nStarting PSO Neural Network Optimization...")

nn_settings = CONFIG["pso_neural_network"]

best_nn_position, best_nn_score, nn_history = (
    particle_swarm_optimization(
        fitness_function=nn_fitness,
        bounds=NN_BOUNDS,
        n_particles=nn_settings["particles"],
        iterations=nn_settings["iterations"],
        inertia=0.7,
        cognitive=1.5,
        social=1.5,
        label="NN-PSO"
    )
)

best_units, best_lr = decode_nn(
    best_nn_position
)

print("\nBest NN Units:", best_units)
print("Best NN Learning Rate:", best_lr)
print("Best NN Validation RMSE:", best_nn_score)

# ==============================================================
# 14. TRAIN PSO NEURAL NETWORK ON TRAIN + VALIDATION
# ==============================================================

final_nn_imputer = SimpleImputer(
    strategy="median"
)

final_nn_scaler = StandardScaler()

X_tv_nn = final_nn_imputer.fit_transform(
    X_train_val
)

X_tv_nn = final_nn_scaler.fit_transform(
    X_tv_nn
)

X_test_nn = final_nn_scaler.transform(
    final_nn_imputer.transform(X_test)
)

tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(SEED)

pso_nn = build_nn(
    best_units,
    best_lr
)

pso_nn.fit(
    X_tv_nn,
    y_train_val.to_numpy(),
    epochs=nn_settings["final_epochs"],
    batch_size=8,
    shuffle=False,
    verbose=0
)

pso_nn_pred = np.maximum(
    pso_nn.predict(
        X_test_nn,
        verbose=0
    ).flatten(),
    0
)

pso_nn_metrics = evaluate(
    y_test,
    pso_nn_pred
)

print("\nPSO Neural Network Performance:")
print(pso_nn_metrics)

# ==============================================================
# 15. MODEL COMPARISON
# ==============================================================

metrics = {
    "Baseline Random Forest": baseline_metrics,
    "PSO Random Forest": pso_rf_metrics,
    "PSO Neural Network": pso_nn_metrics
}

comparison_df = pd.DataFrame(
    metrics
).T

comparison_df.index.name = "Model"

comparison_df = comparison_df.sort_values(
    "RMSE"
)

comparison_df.to_csv(
    ROOT / "pso_model_comparison.csv"
)

print("\nMODEL COMPARISON:")
print(comparison_df)

# Select the forecasting model using validation scores,
# not test scores, to avoid test-set selection leakage.

baseline_validation = create_rf(
    baseline_params
)

baseline_validation.fit(
    X_train,
    y_train
)

baseline_val_score = rmse(
    y_val,
    np.maximum(
        baseline_validation.predict(X_val),
        0
    )
)

validation_scores = {
    "Baseline Random Forest": baseline_val_score,
    "PSO Random Forest": best_rf_score,
    "PSO Neural Network": best_nn_score
}

best_model_name = min(
    validation_scores,
    key=validation_scores.get
)

print("\nSelected Model:", best_model_name)

# ==============================================================
# 16. ACCURACY / PERFORMANCE GRAPH
# ==============================================================

fig, axes = plt.subplots(
    1, 3,
    figsize=(18, 5)
)

metric_names = [
    "R2",
    "MAE",
    "RMSE"
]

titles = [
    "R² Score - Higher is Better",
    "MAE - Lower is Better",
    "RMSE - Lower is Better"
]

for ax, metric, title in zip(
    axes,
    metric_names,
    titles
):

    sns.barplot(
        x=comparison_df.index,
        y=comparison_df[metric].values,
        ax=ax,
        color="teal"
    )

    ax.set_title(title)
    ax.set_xlabel("Models")
    ax.set_ylabel(metric)

    ax.tick_params(
        axis="x",
        rotation=25
    )

    for container in ax.containers:

        ax.bar_label(
            container,
            fmt="%.3f",
            padding=3
        )

plt.suptitle(
    "PSO Rainfall Model Performance",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    ROOT / "pso_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 17. COMPARISON GRAPH
# ==============================================================

fig, ax = plt.subplots(
    figsize=(12, 6)
)

x = np.arange(
    len(comparison_df)
)

width = 0.35

bars1 = ax.bar(
    x - width / 2,
    comparison_df["MAE"],
    width,
    label="MAE"
)

bars2 = ax.bar(
    x + width / 2,
    comparison_df["RMSE"],
    width,
    label="RMSE"
)

ax.set_xticks(x)

ax.set_xticklabels(
    comparison_df.index,
    rotation=15
)

ax.set_title(
    "Baseline vs PSO Optimized Models",
    fontsize=16
)

ax.set_ylabel("Prediction Error")
ax.legend()

ax.bar_label(
    bars1,
    fmt="%.2f",
    padding=3
)

ax.bar_label(
    bars2,
    fmt="%.2f",
    padding=3
)

plt.tight_layout()

plt.savefig(
    ROOT / "pso_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 18. CORRELATION HEATMAP
# ==============================================================

heatmap_columns = [
    "Total_rainfall",
    "month",
    "month_sin",
    "month_cos",
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_6",
    "lag_12",
    "rolling_mean_3",
    "rolling_mean_6",
    "rolling_std_3",
    "rolling_std_6"
]

correlation = featured[
    heatmap_columns
].corr()

plt.figure(
    figsize=(13, 10)
)

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    square=True
)

plt.title(
    "PSO Rainfall Feature Correlation Heatmap",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    ROOT / "pso_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 19. PSO FITNESS CONVERGENCE GRAPH
# ==============================================================

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    range(len(rf_history)),
    rf_history,
    marker="o",
    linewidth=2,
    label="PSO Random Forest"
)

plt.plot(
    range(len(nn_history)),
    nn_history,
    marker="s",
    linewidth=2,
    label="PSO Neural Network"
)

plt.title(
    "PSO Optimization Fitness Convergence",
    fontsize=16
)

plt.xlabel("Iteration")
plt.ylabel("Best Validation RMSE")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    ROOT / "pso_fitness_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 20. RESULT CSV
# ==============================================================

prediction_lookup = {
    "Baseline Random Forest": baseline_pred,
    "PSO Random Forest": pso_rf_pred,
    "PSO Neural Network": pso_nn_pred
}

result_df = pd.DataFrame({
    "Month-Year": test_dates.to_numpy(),
    "Actual_Rainfall": y_test.to_numpy(),
    "Baseline_Prediction": baseline_pred,
    "PSO_RF_Prediction": pso_rf_pred,
    "PSO_NN_Prediction": pso_nn_pred,
    "Best_Model_Prediction":
        prediction_lookup[best_model_name]
})

result_df["Absolute_Error"] = np.abs(
    result_df["Actual_Rainfall"]
    - result_df["Best_Model_Prediction"]
)

result_df["Squared_Error"] = (
    result_df["Actual_Rainfall"]
    - result_df["Best_Model_Prediction"]
) ** 2

result_df["Month-Year"] = (
    result_df["Month-Year"].dt.strftime(
        "%Y-%m"
    )
)

result_df.to_csv(
    ROOT / "pso_result.csv",
    index=False
)

print("\nPSO RESULT CSV:")
print(result_df.head())

# ==============================================================
# 21. RESULT GRAPH
# ==============================================================

plt.figure(
    figsize=(15, 7)
)

plt.plot(
    test_dates,
    y_test,
    label="Actual Rainfall",
    color="black",
    linewidth=2.5,
    marker="o"
)

plt.plot(
    test_dates,
    baseline_pred,
    label="Baseline Random Forest",
    linestyle="--",
    marker="s"
)

plt.plot(
    test_dates,
    pso_rf_pred,
    label="PSO Random Forest",
    linestyle="--",
    marker="^"
)

plt.plot(
    test_dates,
    pso_nn_pred,
    label="PSO Neural Network",
    linestyle=":",
    marker="d"
)

plt.title(
    "PSO Actual vs Predicted Monthly Rainfall",
    fontsize=16
)

plt.xlabel("Month-Year")
plt.ylabel("Total Rainfall")

plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    ROOT / "pso_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 22. FEATURE IMPORTANCE GRAPH
# ==============================================================

# Fit the optimized RF on all historical samples
# for final forecasting and feature importance.

final_rf = create_rf(
    best_rf_params
)

final_rf.fit(X, y)

importance = final_rf.named_steps[
    "model"
].feature_importances_

importance_df = pd.DataFrame({
    "Feature": FEATURE_COLUMNS,
    "Importance": importance
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

importance_df.to_csv(
    ROOT / "pso_feature_importance.csv",
    index=False
)

plt.figure(
    figsize=(12, 7)
)

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature",
    color="teal"
)

plt.title(
    "PSO Random Forest Feature Importance",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    ROOT / "pso_feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 23. REFIT FINAL NEURAL NETWORK ON ALL DATA
# ==============================================================

all_nn_imputer = SimpleImputer(
    strategy="median"
)

all_nn_scaler = StandardScaler()

X_all_nn = all_nn_imputer.fit_transform(X)

X_all_nn = all_nn_scaler.fit_transform(
    X_all_nn
)

tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(SEED)

final_nn = build_nn(
    best_units,
    best_lr
)

final_nn.fit(
    X_all_nn,
    y.to_numpy(),
    epochs=nn_settings["final_epochs"],
    batch_size=8,
    shuffle=False,
    verbose=0
)

final_baseline = create_rf(
    baseline_params
)

final_baseline.fit(X, y)

# ==============================================================
# 24. SAVE PKL AND H5 MODELS
# ==============================================================

PKL_PATH = ROOT / "pso_rainfall_model.pkl"

joblib.dump(
    {
        "model": final_rf,
        "model_name": "PSO Random Forest",
        "feature_columns": FEATURE_COLUMNS,
        "best_parameters": best_rf_params,
        "nn_imputer": all_nn_imputer,
        "nn_scaler": all_nn_scaler,
        "selected_best_model": best_model_name,
        "baseline_model": final_baseline
    },
    PKL_PATH
)

H5_PATH = ROOT / "pso_rainfall_model.h5"

final_nn.save(
    str(H5_PATH)
)

print("\nSaved PSO PKL:", PKL_PATH)
print("Saved PSO H5:", H5_PATH)

# ==============================================================
# 25. FUTURE FEATURE CREATION
# ==============================================================

def build_future_features(
    history,
    next_date
):

    future_row = pd.DataFrame({
        "Month-Year": [next_date],
        "Total_rainfall": [np.nan]
    })

    temp = pd.concat(
        [history, future_row],
        ignore_index=True
    )

    temp_features = create_features(temp)

    return temp_features[
        FEATURE_COLUMNS
    ].iloc[[-1]]

def predict_future(features):

    if best_model_name == "PSO Neural Network":

        transformed = all_nn_imputer.transform(
            features
        )

        transformed = all_nn_scaler.transform(
            transformed
        )

        prediction = final_nn.predict(
            transformed,
            verbose=0
        ).flatten()[0]

    elif best_model_name == "PSO Random Forest":

        prediction = final_rf.predict(
            features
        )[0]

    else:

        prediction = final_baseline.predict(
            features
        )[0]

    return max(
        float(prediction),
        0.0
    )

# ==============================================================
# 26. FUTURE 12-MONTH RAINFALL FORECAST
# ==============================================================

forecast_history = df.copy()

future_records = []

FORECAST_MONTHS = CONFIG["forecast_months"]

for step in range(FORECAST_MONTHS):

    next_date = (
        forecast_history[
            "Month-Year"
        ].iloc[-1]
        + pd.DateOffset(months=1)
    )

    future_X = build_future_features(
        forecast_history,
        next_date
    )

    predicted_rainfall = predict_future(
        future_X
    )

    future_records.append({
        "Month-Year": next_date,
        "Predicted_Rainfall": round(
            predicted_rainfall,
            3
        ),
        "Model_Used": best_model_name
    })

    forecast_history = pd.concat([
        forecast_history,
        pd.DataFrame({
            "Month-Year": [next_date],
            "Total_rainfall": [
                predicted_rainfall
            ]
        })
    ], ignore_index=True)

prediction_df = pd.DataFrame(
    future_records
)

prediction_df.to_csv(
    ROOT / "pso_prediction.csv",
    index=False,
    date_format="%Y-%m"
)

print("\nPSO FUTURE RAINFALL PREDICTIONS:")
print(prediction_df)

# ==============================================================
# 27. FUTURE PREDICTION GRAPH
# ==============================================================

historical_plot = df.tail(36)

plt.figure(
    figsize=(15, 7)
)

plt.plot(
    historical_plot["Month-Year"],
    historical_plot["Total_rainfall"],
    label="Historical Rainfall",
    color="royalblue",
    linewidth=2,
    marker="o"
)

plt.plot(
    prediction_df["Month-Year"],
    prediction_df["Predicted_Rainfall"],
    label="PSO Forecast",
    color="crimson",
    linewidth=2.5,
    marker="s",
    linestyle="--"
)

plt.plot(
    [
        historical_plot["Month-Year"].iloc[-1],
        prediction_df["Month-Year"].iloc[0]
    ],
    [
        historical_plot["Total_rainfall"].iloc[-1],
        prediction_df["Predicted_Rainfall"].iloc[0]
    ],
    color="crimson",
    linestyle="--"
)

plt.axvline(
    historical_plot["Month-Year"].iloc[-1],
    color="gray",
    linestyle=":",
    label="Forecast Start"
)

plt.title(
    "PSO Monthly Rainfall Forecast - Next 12 Months",
    fontsize=16
)

plt.xlabel("Month-Year")
plt.ylabel("Total Rainfall")

plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    ROOT / "pso_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 28. SAVE MODEL INFORMATION JSON
# ==============================================================

model_info = {
    "project": CONFIG["project"],
    "algorithm": CONFIG["algorithm"],
    "created_at": datetime.now().isoformat(),
    "dataset": CSV_PATH.name,
    "total_records": int(len(df)),
    "training_samples": int(len(X_train)),
    "validation_samples": int(len(X_val)),
    "testing_samples": int(len(X_test)),
    "feature_columns": FEATURE_COLUMNS,
    "best_rf_parameters": best_rf_params,
    "best_nn_units": best_units,
    "best_nn_learning_rate": best_lr,
    "best_rf_validation_rmse": float(best_rf_score),
    "best_nn_validation_rmse": float(best_nn_score),
    "baseline_validation_rmse": float(
        baseline_val_score
    ),
    "selected_best_model": best_model_name,
    "selection_method": "Minimum validation RMSE",
    "model_metrics": metrics,
    "forecast_months": FORECAST_MONTHS,
    "saved_files": [
        "pso_rainfall_model.pkl",
        "pso_rainfall_model.h5",
        "pso_config.yaml",
        "pso_model_info.json",
        "pso_accuracy_graph.png",
        "pso_comparison_graph.png",
        "pso_heatmap.png",
        "pso_result.csv",
        "pso_result_graph.png",
        "pso_prediction.csv",
        "pso_prediction_graph.png",
        "pso_fitness_graph.png",
        "pso_model_comparison.csv",
        "pso_feature_importance.csv",
        "pso_feature_importance.png"
    ]
}

with open(
    ROOT / "pso_model_info.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        model_info,
        f,
        indent=4
    )

# ==============================================================
# 29. FINAL OUTPUT SUMMARY
# ==============================================================

print("\n" + "=" * 65)
print("PSO RAINFALL FORECASTING COMPLETED")
print("=" * 65)

print("\nSelected Model:", best_model_name)

print("\nMODEL PERFORMANCE:")

for name, metric in metrics.items():

    print(
        f"{name:25s} | "
        f"MAE={metric['MAE']:.4f} | "
        f"RMSE={metric['RMSE']:.4f} | "
        f"R2={metric['R2']:.4f}"
    )

print("\nGENERATED FILES:")

for filename in model_info["saved_files"]:

    path = ROOT / filename

    if path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nOutput Directory:")
print(ROOT)

print("\nALL PSO RESULTS GENERATED SUCCESSFULLY!")
```
