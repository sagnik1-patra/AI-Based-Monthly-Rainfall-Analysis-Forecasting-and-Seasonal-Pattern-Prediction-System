```python
import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import tensorflow as tf

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT PATH
# ============================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads\AI-Based Monthly Rainfall Analysis, Forecasting, and Seasonal Pattern Prediction System"
)

CSV_PATH = ROOT / "D67. Rainfall Monthly Total_1_0.csv"
PKL_PATH = ROOT / "rainfall_model.pkl"
H5_PATH = ROOT / "rainfall_model.h5"
JSON_PATH = ROOT / "model_info.json"

ROOT.mkdir(parents=True, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

print("=" * 65)
print("AI-BASED MONTHLY RAINFALL FORECASTING SYSTEM")
print("=" * 65)

# ============================================================
# 2. LOAD DATASET
# ============================================================

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

required_columns = [
    "Month-Year",
    "Total_rainfall"
]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(
            f"Required column missing: {col}"
        )

df["Month-Year"] = pd.to_datetime(
    df["Month-Year"],
    format="%b-%y",
    errors="coerce"
)

df["Total_rainfall"] = pd.to_numeric(
    df["Total_rainfall"],
    errors="coerce"
)

df = df.dropna(
    subset=["Month-Year", "Total_rainfall"]
)

df = df.sort_values(
    "Month-Year"
).reset_index(drop=True)

if df["Month-Year"].duplicated().any():
    raise ValueError("Duplicate months found.")

if (df["Total_rainfall"] < 0).any():
    raise ValueError("Negative rainfall values found.")

expected_months = pd.date_range(
    df["Month-Year"].min(),
    df["Month-Year"].max(),
    freq="MS"
)

if not np.array_equal(
    expected_months.to_numpy(),
    df["Month-Year"].to_numpy()
):
    raise ValueError(
        "Dataset must have continuous monthly records."
    )

print("\nDataset loaded successfully!")
print(df.head())
print("\nDataset shape:", df.shape)

# ============================================================
# 3. FEATURE ENGINEERING
# ============================================================

LAGS = [1, 2, 3, 6, 12]
WINDOWS = [3, 6]

FEATURE_COLUMNS = [
    "month",
    "month_sin",
    "month_cos",
    "time_index"
]

FEATURE_COLUMNS += [
    f"lag_{lag}" for lag in LAGS
]

FEATURE_COLUMNS += [
    f"rolling_mean_{w}" for w in WINDOWS
]

FEATURE_COLUMNS += [
    f"rolling_std_{w}" for w in WINDOWS
]

def create_features(data):

    result = data.copy()

    result["month"] = (
        result["Month-Year"].dt.month
    )

    result["month_sin"] = np.sin(
        2 * np.pi * result["month"] / 12
    )

    result["month_cos"] = np.cos(
        2 * np.pi * result["month"] / 12
    )

    result["time_index"] = np.arange(
        len(result)
    )

    for lag in LAGS:
        result[f"lag_{lag}"] = (
            result["Total_rainfall"].shift(lag)
        )

    previous = result[
        "Total_rainfall"
    ].shift(1)

    for window in WINDOWS:

        result[f"rolling_mean_{window}"] = (
            previous.rolling(window).mean()
        )

        result[f"rolling_std_{window}"] = (
            previous.rolling(window).std()
        )

    return result

featured = create_features(df)

featured = featured.dropna().reset_index(
    drop=True
)

X = featured[FEATURE_COLUMNS]
y = featured["Total_rainfall"]

print("\nFeature engineering completed.")
print("Features:", FEATURE_COLUMNS)

# ============================================================
# 4. TRAIN TEST SPLIT
# ============================================================

split_index = int(len(X) * 0.80)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

test_dates = featured[
    "Month-Year"
].iloc[split_index:]

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

# ============================================================
# 5. LOAD / TRAIN RANDOM FOREST AND GRADIENT BOOSTING
# ============================================================

models = {
    "Random Forest": Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                max_depth=8,
                min_samples_leaf=2,
                random_state=SEED,
                n_jobs=-1
            )
        )
    ]),

    "Gradient Boosting": Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=150,
                learning_rate=0.05,
                max_depth=2,
                random_state=SEED
            )
        )
    ])
}

results = {}
predictions = {}

def evaluate_model(actual, predicted):

    mae = mean_absolute_error(
        actual, predicted
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual, predicted
        )
    )

    r2 = r2_score(
        actual, predicted
    )

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2)
    }

for name, model in models.items():

    print(f"\nTraining {name}...")

    model.fit(X_train, y_train)

    pred = model.predict(X_test)

    pred = np.maximum(pred, 0)

    predictions[name] = pred

    results[name] = evaluate_model(
        y_test, pred
    )

    print(name, results[name])

# ============================================================
# 6. NEURAL NETWORK EVALUATION
# ============================================================

# The H5 model from the previous training script
# was fitted on all available data.
# Evaluating it on X_test would cause data leakage.
#
# Therefore, train a fresh evaluation-only neural
# network on the training period.

nn_imputer = SimpleImputer(
    strategy="median"
)

nn_scaler = StandardScaler()

X_train_nn = nn_imputer.fit_transform(
    X_train
)

X_train_nn = nn_scaler.fit_transform(
    X_train_nn
)

X_test_nn = nn_imputer.transform(
    X_test
)

X_test_nn = nn_scaler.transform(
    X_test_nn
)

validation_count = max(
    8,
    int(len(X_train_nn) * 0.20)
)

X_fit = X_train_nn[:-validation_count]
X_val = X_train_nn[-validation_count:]

y_fit = y_train.to_numpy()[
    :-validation_count
]

y_val = y_train.to_numpy()[
    -validation_count:
]

tf.keras.backend.clear_session()
tf.random.set_seed(SEED)

nn = tf.keras.Sequential([
    tf.keras.layers.Input(
        shape=(X_train_nn.shape[1],)
    ),
    tf.keras.layers.Dense(
        64, activation="relu"
    ),
    tf.keras.layers.Dropout(0.15),
    tf.keras.layers.Dense(
        32, activation="relu"
    ),
    tf.keras.layers.Dense(
        16, activation="relu"
    ),
    tf.keras.layers.Dense(1)
])

nn.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=20,
    restore_best_weights=True
)

print("\nTraining Neural Network...")

history = nn.fit(
    X_fit,
    y_fit,
    validation_data=(X_val, y_val),
    epochs=200,
    batch_size=8,
    callbacks=[early_stop],
    verbose=0,
    shuffle=False
)

nn_pred = nn.predict(
    X_test_nn,
    verbose=0
).flatten()

nn_pred = np.maximum(
    nn_pred, 0
)

predictions["Neural Network"] = nn_pred

results["Neural Network"] = evaluate_model(
    y_test, nn_pred
)

print(
    "Neural Network:",
    results["Neural Network"]
)

# ============================================================
# 7. MODEL COMPARISON DATAFRAME
# ============================================================

comparison_df = pd.DataFrame(
    results
).T

comparison_df.index.name = "Model"

comparison_df = comparison_df.sort_values(
    "RMSE"
)

comparison_df.to_csv(
    ROOT / "model_comparison.csv"
)

print("\nMODEL COMPARISON")
print(comparison_df)

best_model_name = comparison_df.index[0]

print("\nBest Model:", best_model_name)

# ============================================================
# 8. ACCURACY / PERFORMANCE GRAPH
# ============================================================

# Rainfall prediction is regression.
# R2, MAE and RMSE are more appropriate than
# classification accuracy.

fig, axes = plt.subplots(
    1, 3,
    figsize=(18, 5)
)

metrics = ["R2", "MAE", "RMSE"]

titles = [
    "R² Score (Higher is Better)",
    "Mean Absolute Error (Lower is Better)",
    "Root Mean Squared Error (Lower is Better)"
]

for ax, metric, title in zip(
    axes, metrics, titles
):

    sns.barplot(
        x=comparison_df.index,
        y=comparison_df[metric].values,
        ax=ax,
        color="steelblue"
    )

    ax.set_title(
        title,
        fontsize=12,
        fontweight="bold"
    )

    ax.set_xlabel("Models")
    ax.set_ylabel(metric)

    ax.tick_params(
        axis="x",
        rotation=25
    )

    for container in ax.containers:
        ax.bar_label(
            container,
            fmt="%.3f",
            padding=3
        )

plt.suptitle(
    "Rainfall Prediction Model Performance",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    ROOT / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved accuracy_graph.png")

# ============================================================
# 9. MODEL COMPARISON GRAPH
# ============================================================

fig, ax = plt.subplots(
    figsize=(12, 6)
)

x = np.arange(
    len(comparison_df)
)

width = 0.35

bars1 = ax.bar(
    x - width / 2,
    comparison_df["MAE"],
    width,
    label="MAE",
    color="royalblue"
)

bars2 = ax.bar(
    x + width / 2,
    comparison_df["RMSE"],
    width,
    label="RMSE",
    color="darkorange"
)

ax.set_xticks(x)
ax.set_xticklabels(
    comparison_df.index,
    rotation=15
)

ax.set_title(
    "Machine Learning Model Comparison",
    fontsize=15,
    fontweight="bold"
)

ax.set_ylabel("Prediction Error")
ax.set_xlabel("Machine Learning Models")

ax.legend()

ax.bar_label(
    bars1,
    fmt="%.2f",
    padding=3
)

ax.bar_label(
    bars2,
    fmt="%.2f",
    padding=3
)

plt.tight_layout()

plt.savefig(
    ROOT / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved comparison_graph.png")

# ============================================================
# 10. CORRELATION HEATMAP
# ============================================================

heatmap_columns = [
    "Total_rainfall",
    "month",
    "month_sin",
    "month_cos",
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_6",
    "lag_12",
    "rolling_mean_3",
    "rolling_mean_6",
    "rolling_std_3",
    "rolling_std_6"
]

correlation = featured[
    heatmap_columns
].corr()

plt.figure(
    figsize=(13, 10)
)

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    square=True
)

plt.title(
    "Rainfall Feature Correlation Heatmap",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    ROOT / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved heatmap.png")

# ============================================================
# 11. RESULT CSV
# ============================================================

result_df = pd.DataFrame({
    "Month-Year": test_dates.to_numpy(),
    "Actual_Rainfall": y_test.to_numpy(),
    "Random_Forest_Prediction":
        predictions["Random Forest"],
    "Gradient_Boosting_Prediction":
        predictions["Gradient Boosting"],
    "Neural_Network_Prediction":
        predictions["Neural Network"]
})

result_df["Best_Model_Prediction"] = (
    predictions[best_model_name]
)

result_df["Absolute_Error"] = np.abs(
    result_df["Actual_Rainfall"]
    - result_df["Best_Model_Prediction"]
)

result_df["Squared_Error"] = (
    result_df["Actual_Rainfall"]
    - result_df["Best_Model_Prediction"]
) ** 2

result_df["Month-Year"] = (
    result_df["Month-Year"].dt.strftime(
        "%Y-%m"
    )
)

result_df.to_csv(
    ROOT / "result.csv",
    index=False
)

print("\nRESULT CSV")
print(result_df.head(10))

print("\nSaved result.csv")

# ============================================================
# 12. RESULT GRAPH
# ============================================================

plt.figure(
    figsize=(15, 7)
)

plt.plot(
    test_dates,
    y_test,
    label="Actual Rainfall",
    color="black",
    marker="o",
    linewidth=2.5
)

plt.plot(
    test_dates,
    predictions["Random Forest"],
    label="Random Forest",
    linestyle="--",
    marker="s"
)

plt.plot(
    test_dates,
    predictions["Gradient Boosting"],
    label="Gradient Boosting",
    linestyle="--",
    marker="^"
)

plt.plot(
    test_dates,
    predictions["Neural Network"],
    label="Neural Network",
    linestyle=":",
    marker="d"
)

plt.title(
    "Actual vs Predicted Monthly Rainfall",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Month-Year")
plt.ylabel("Total Rainfall")

plt.legend()
plt.grid(True, alpha=0.3)

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    ROOT / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved result_graph.png")

# ============================================================
# 13. LOAD FINAL SAVED MODELS
# ============================================================

if not PKL_PATH.exists():
    raise FileNotFoundError(
        "rainfall_model.pkl not found. "
        "Run the previous training code first."
    )

if not H5_PATH.exists():
    raise FileNotFoundError(
        "rainfall_model.h5 not found. "
        "Run the previous training code first."
    )

bundle = joblib.load(PKL_PATH)

saved_classical_model = bundle["model"]
saved_classical_name = bundle["model_name"]

saved_nn = tf.keras.models.load_model(
    str(H5_PATH),
    compile=False
)

saved_nn_imputer = bundle[
    "neural_network_imputer"
]

saved_nn_scaler = bundle[
    "neural_network_scaler"
]

# For consistency with the new comparison,
# choose the best model by the current test RMSE.

if best_model_name == "Neural Network":
    forecast_model = saved_nn
else:
    # The PKL stores only the best classical model.
    # Refit the chosen classical model on all data
    # if it differs from the saved classical model.
    if best_model_name == saved_classical_name:
        forecast_model = saved_classical_model
    else:
        forecast_model = models[best_model_name]
        forecast_model.fit(X, y)

print("\nForecasting model:", best_model_name)

# ============================================================
# 14. FUTURE FEATURE CREATION
# ============================================================

def build_future_features(
    history_df,
    next_date
):

    next_row = pd.DataFrame({
        "Month-Year": [next_date],
        "Total_rainfall": [np.nan]
    })

    temp = pd.concat(
        [history_df, next_row],
        ignore_index=True
    )

    temp_features = create_features(temp)

    return temp_features[
        FEATURE_COLUMNS
    ].iloc[[-1]]

def predict_future_rainfall(
    model_name,
    features
):

    if model_name == "Neural Network":

        scaled = saved_nn_imputer.transform(
            features
        )

        scaled = saved_nn_scaler.transform(
            scaled
        )

        prediction = forecast_model.predict(
            scaled,
            verbose=0
        ).flatten()[0]

    else:

        prediction = forecast_model.predict(
            features
        )[0]

    return max(
        float(prediction),
        0.0
    )

# ============================================================
# 15. PREDICT NEXT 12 MONTHS
# ============================================================

forecast_history = df.copy()

future_records = []

FORECAST_MONTHS = 12

for i in range(FORECAST_MONTHS):

    next_date = (
        forecast_history[
            "Month-Year"
        ].iloc[-1]
        + pd.DateOffset(months=1)
    )

    future_features = build_future_features(
        forecast_history,
        next_date
    )

    predicted_rainfall = (
        predict_future_rainfall(
            best_model_name,
            future_features
        )
    )

    future_records.append({
        "Month-Year": next_date,
        "Predicted_Rainfall": round(
            predicted_rainfall,
            3
        ),
        "Model_Used": best_model_name
    })

    new_row = pd.DataFrame({
        "Month-Year": [next_date],
        "Total_rainfall": [
            predicted_rainfall
        ]
    })

    forecast_history = pd.concat(
        [forecast_history, new_row],
        ignore_index=True
    )

prediction_df = pd.DataFrame(
    future_records
)

prediction_df.to_csv(
    ROOT / "prediction.csv",
    index=False,
    date_format="%Y-%m"
)

print("\nFUTURE RAINFALL PREDICTIONS")
print(prediction_df)

print("\nSaved prediction.csv")

# ============================================================
# 16. PREDICTION GRAPH
# ============================================================

# Display the last 36 historical months
# and the next 12 predicted months.

historical_plot = df.tail(36)

plt.figure(
    figsize=(15, 7)
)

plt.plot(
    historical_plot["Month-Year"],
    historical_plot["Total_rainfall"],
    label="Historical Rainfall",
    color="royalblue",
    linewidth=2,
    marker="o"
)

plt.plot(
    prediction_df["Month-Year"],
    prediction_df["Predicted_Rainfall"],
    label="Forecasted Rainfall",
    color="crimson",
    linewidth=2.5,
    marker="s",
    linestyle="--"
)

# Connect historical series to forecast.
plt.plot(
    [
        historical_plot["Month-Year"].iloc[-1],
        prediction_df["Month-Year"].iloc[0]
    ],
    [
        historical_plot["Total_rainfall"].iloc[-1],
        prediction_df["Predicted_Rainfall"].iloc[0]
    ],
    color="crimson",
    linestyle="--",
    alpha=0.6
)

plt.axvline(
    x=historical_plot["Month-Year"].iloc[-1],
    color="gray",
    linestyle=":",
    label="Forecast Start"
)

plt.title(
    "Monthly Rainfall Forecast - Next 12 Months",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Month-Year")
plt.ylabel("Total Rainfall")

plt.legend()
plt.grid(True, alpha=0.3)
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    ROOT / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved prediction_graph.png")

# ============================================================
# 17. SAVE UPDATED MODEL COMPARISON METADATA
# ============================================================

metadata = {
    "project": (
        "AI-Based Monthly Rainfall Analysis, "
        "Forecasting, and Seasonal Pattern Prediction"
    ),
    "best_model_by_test_rmse": best_model_name,
    "evaluation": "chronological holdout",
    "training_samples": int(len(X_train)),
    "testing_samples": int(len(X_test)),
    "forecast_months": FORECAST_MONTHS,
    "metrics": results,
    "outputs": [
        "accuracy_graph.png",
        "comparison_graph.png",
        "heatmap.png",
        "result.csv",
        "result_graph.png",
        "prediction.csv",
        "prediction_graph.png",
        "model_comparison.csv"
    ]
}

with open(
    ROOT / "visualization_results.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        metadata,
        f,
        indent=4
    )

# ============================================================
# 18. FINAL OUTPUT SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("ALL RESULTS GENERATED SUCCESSFULLY")
print("=" * 65)

output_files = [
    "accuracy_graph.png",
    "comparison_graph.png",
    "heatmap.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "model_comparison.csv",
    "visualization_results.json"
]

for filename in output_files:

    path = ROOT / filename

    if path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nBest Model:", best_model_name)

print("\nMODEL PERFORMANCE:")

for name, metric in results.items():

    print(
        f"{name:20s} | "
        f"MAE: {metric['MAE']:.4f} | "
        f"RMSE: {metric['RMSE']:.4f} | "
        f"R2: {metric['R2']:.4f}"
    )

print("\nOutput directory:")
print(ROOT)

print("\nPROJECT VISUALIZATION COMPLETED!")
```
