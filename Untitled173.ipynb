```python
from pathlib import Path
import json
import random
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
import joblib
import tensorflow as tf

from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ==============================================================
# 1. PROJECT CONFIGURATION
# ==============================================================

ROOT = Path(
    r"C:\Users\sagni\Downloads\AI-Based Monthly Rainfall Analysis, Forecasting, and Seasonal Pattern Prediction System"
)

ROOT.mkdir(parents=True, exist_ok=True)

CSV_PATH = ROOT / "D67. Rainfall Monthly Total_1_0.csv"

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

CONFIG = {
    "project": "AIS-Based Monthly Rainfall Forecasting",
    "algorithm": "Artificial Immune System - CLONALG",
    "dataset": str(CSV_PATH),
    "seed": SEED,
    "forecast_months": 12,
    "lags": [1, 2, 3, 6, 12],
    "rolling_windows": [3, 6],
    "train_ratio": 0.60,
    "validation_ratio": 0.20,
    "test_ratio": 0.20,
    "ais": {
        "population_size": 10,
        "generations": 12,
        "elite_count": 3,
        "clones_per_elite": 4,
        "random_immigrants": 2,
        "mutation_strength": 0.20
    },
    "neural_network": {
        "epochs": 100,
        "batch_size": 8
    }
}

with open(
    ROOT / "ais_config.yaml",
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(CONFIG, f, sort_keys=False)

print("=" * 65)
print("AIS-BASED RAINFALL FORECASTING SYSTEM")
print("=" * 65)

# ==============================================================
# 2. LOAD DATA
# ==============================================================

if not CSV_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

required = ["Month-Year", "Total_rainfall"]

for column in required:
    if column not in df.columns:
        raise ValueError(
            f"Missing required column: {column}"
        )

df["Month-Year"] = pd.to_datetime(
    df["Month-Year"],
    format="%b-%y",
    errors="coerce"
)

df["Total_rainfall"] = pd.to_numeric(
    df["Total_rainfall"],
    errors="coerce"
)

df = df.dropna(
    subset=required
).sort_values(
    "Month-Year"
).reset_index(drop=True)

if df["Month-Year"].duplicated().any():
    raise ValueError("Duplicate months found.")

if (df["Total_rainfall"] < 0).any():
    raise ValueError("Negative rainfall values found.")

expected = pd.date_range(
    df["Month-Year"].min(),
    df["Month-Year"].max(),
    freq="MS"
)

if not np.array_equal(
    expected.to_numpy(),
    df["Month-Year"].to_numpy()
):
    raise ValueError(
        "Monthly observations must be continuous."
    )

print("\nDataset loaded successfully")
print(df.head())
print("Dataset shape:", df.shape)

# ==============================================================
# 3. FEATURE ENGINEERING
# ==============================================================

LAGS = CONFIG["lags"]
WINDOWS = CONFIG["rolling_windows"]

FEATURE_COLUMNS = [
    "month",
    "month_sin",
    "month_cos",
    "time_index"
]

FEATURE_COLUMNS += [
    f"lag_{lag}" for lag in LAGS
]

FEATURE_COLUMNS += [
    f"rolling_mean_{w}" for w in WINDOWS
]

FEATURE_COLUMNS += [
    f"rolling_std_{w}" for w in WINDOWS
]

def create_features(data):

    result = data.copy()

    result["month"] = result["Month-Year"].dt.month

    result["month_sin"] = np.sin(
        2 * np.pi * result["month"] / 12
    )

    result["month_cos"] = np.cos(
        2 * np.pi * result["month"] / 12
    )

    result["time_index"] = np.arange(
        len(result)
    )

    for lag in LAGS:
        result[f"lag_{lag}"] = (
            result["Total_rainfall"].shift(lag)
        )

    previous = result["Total_rainfall"].shift(1)

    for window in WINDOWS:

        result[f"rolling_mean_{window}"] = (
            previous.rolling(window).mean()
        )

        result[f"rolling_std_{window}"] = (
            previous.rolling(window).std()
        )

    return result

featured = create_features(df)

featured = featured.dropna().reset_index(drop=True)

X = featured[FEATURE_COLUMNS]
y = featured["Total_rainfall"]

dates = featured["Month-Year"]

print("\nFeature engineering completed")
print("Features:", FEATURE_COLUMNS)

# ==============================================================
# 4. CHRONOLOGICAL TRAIN / VALIDATION / TEST SPLIT
# ==============================================================

n = len(X)

train_end = int(n * 0.60)
validation_end = int(n * 0.80)

X_train = X.iloc[:train_end]
y_train = y.iloc[:train_end]

X_val = X.iloc[train_end:validation_end]
y_val = y.iloc[train_end:validation_end]

X_test = X.iloc[validation_end:]
y_test = y.iloc[validation_end:]

test_dates = dates.iloc[validation_end:]

X_train_val = X.iloc[:validation_end]
y_train_val = y.iloc[:validation_end]

print("\nTraining samples:", len(X_train))
print("Validation samples:", len(X_val))
print("Testing samples:", len(X_test))

# ==============================================================
# 5. EVALUATION FUNCTION
# ==============================================================

def evaluate(actual, predicted):

    predicted = np.maximum(
        np.asarray(predicted).flatten(),
        0
    )

    return {
        "MAE": float(
            mean_absolute_error(actual, predicted)
        ),
        "RMSE": float(
            np.sqrt(
                mean_squared_error(actual, predicted)
            )
        ),
        "R2": float(
            r2_score(actual, predicted)
        )
    }

# ==============================================================
# 6. BASELINE RANDOM FOREST
# ==============================================================

baseline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        RandomForestRegressor(
            n_estimators=100,
            max_depth=8,
            min_samples_leaf=2,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

baseline.fit(X_train_val, y_train_val)

baseline_pred = np.maximum(
    baseline.predict(X_test),
    0
)

baseline_metrics = evaluate(
    y_test,
    baseline_pred
)

print("\nBaseline Random Forest:")
print(baseline_metrics)

# ==============================================================
# 7. AIS PARAMETER SEARCH SPACE
# ==============================================================

# Each antibody is a normalized vector of 4 values.
#
# 0: n_estimators
# 1: max_depth
# 2: min_samples_split
# 3: min_samples_leaf

BOUNDS = {
    "n_estimators": (50, 250),
    "max_depth": (2, 15),
    "min_samples_split": (2, 12),
    "min_samples_leaf": (1, 6)
}

PARAMETER_NAMES = list(BOUNDS.keys())

def decode_antibody(antibody):

    params = {}

    for i, name in enumerate(PARAMETER_NAMES):

        low, high = BOUNDS[name]

        value = low + antibody[i] * (
            high - low
        )

        params[name] = int(
            round(value)
        )

    return params

def random_antibody():

    return np.random.uniform(
        0, 1,
        size=len(PARAMETER_NAMES)
    )

def create_rf(params):

    return Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "model",
            RandomForestRegressor(
                **params,
                random_state=SEED,
                n_jobs=-1
            )
        )
    ])

# ==============================================================
# 8. AIS FITNESS FUNCTION
# ==============================================================

fitness_cache = {}

def fitness(antibody):

    params = decode_antibody(antibody)

    key = tuple(
        params[name]
        for name in PARAMETER_NAMES
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = create_rf(params)

    model.fit(
        X_train,
        y_train
    )

    pred = np.maximum(
        model.predict(X_val),
        0
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_val,
            pred
        )
    )

    fitness_cache[key] = float(rmse)

    return float(rmse)

# ==============================================================
# 9. CLONAL SELECTION ALGORITHM
# ==============================================================

AIS_CONFIG = CONFIG["ais"]

POP_SIZE = AIS_CONFIG["population_size"]
GENERATIONS = AIS_CONFIG["generations"]
ELITE_COUNT = AIS_CONFIG["elite_count"]
CLONES = AIS_CONFIG["clones_per_elite"]
IMMIGRANTS = AIS_CONFIG["random_immigrants"]
MUTATION_STRENGTH = AIS_CONFIG["mutation_strength"]

population = [
    random_antibody()
    for _ in range(POP_SIZE)
]

best_antibody = None
best_fitness = float("inf")

fitness_history = []

print("\nStarting AIS optimization...")

for generation in range(GENERATIONS):

    scored = [
        (fitness(antibody), antibody.copy())
        for antibody in population
    ]

    scored.sort(
        key=lambda item: item[0]
    )

    generation_best = scored[0][0]

    if generation_best < best_fitness:

        best_fitness = generation_best

        best_antibody = scored[0][1].copy()

    fitness_history.append(best_fitness)

    print(
        f"Generation {generation + 1:02d}/"
        f"{GENERATIONS} | "
        f"Best Validation RMSE: "
        f"{best_fitness:.4f}"
    )

    elites = [
        antibody.copy()
        for _, antibody in scored[:ELITE_COUNT]
    ]

    clone_population = []

    for rank, elite in enumerate(elites):

        # Better antibodies receive smaller mutations.
        mutation_scale = (
            MUTATION_STRENGTH
            * (rank + 1)
            / ELITE_COUNT
        )

        for _ in range(CLONES):

            clone = elite.copy()

            mutation = np.random.normal(
                loc=0,
                scale=mutation_scale,
                size=len(clone)
            )

            clone = np.clip(
                clone + mutation,
                0,
                1
            )

            clone_population.append(clone)

    candidates = (
        elites
        + clone_population
        + [
            random_antibody()
            for _ in range(IMMIGRANTS)
        ]
    )

    candidate_scores = [
        (fitness(candidate), candidate)
        for candidate in candidates
    ]

    candidate_scores.sort(
        key=lambda item: item[0]
    )

    population = [
        antibody.copy()
        for _, antibody in candidate_scores[
            :POP_SIZE - IMMIGRANTS
        ]
    ]

    population += [
        random_antibody()
        for _ in range(IMMIGRANTS)
    ]

best_params = decode_antibody(
    best_antibody
)

print("\nAIS Optimization Completed!")
print("Best Parameters:", best_params)
print("Best Validation RMSE:", best_fitness)

# ==============================================================
# 10. TRAIN FINAL AIS RANDOM FOREST
# ==============================================================

ais_rf = create_rf(best_params)

ais_rf.fit(
    X_train_val,
    y_train_val
)

ais_rf_pred = np.maximum(
    ais_rf.predict(X_test),
    0
)

ais_rf_metrics = evaluate(
    y_test,
    ais_rf_pred
)

print("\nAIS Random Forest Performance:")
print(ais_rf_metrics)

# ==============================================================
# 11. AIS-OPTIMIZED NEURAL NETWORK
# ==============================================================

# A second AIS search tunes the neural network's
# hidden-layer width and learning rate.
# This creates a genuine AIS-optimized H5 model.

nn_imputer = SimpleImputer(
    strategy="median"
)

nn_scaler = StandardScaler()

X_train_nn = nn_imputer.fit_transform(
    X_train
)

X_train_nn = nn_scaler.fit_transform(
    X_train_nn
)

X_val_nn = nn_scaler.transform(
    nn_imputer.transform(X_val)
)

X_test_nn = nn_scaler.transform(
    nn_imputer.transform(X_test)
)

X_train_val_nn = nn_scaler.transform(
    nn_imputer.transform(X_train_val)
)

def decode_nn_antibody(antibody):

    units = int(
        round(
            16 + antibody[0] * 112
        )
    )

    learning_rate = float(
        10 ** (
            -4 + antibody[1] * 2
        )
    )

    return units, learning_rate

def build_nn(units, learning_rate):

    model = tf.keras.Sequential([
        tf.keras.layers.Input(
            shape=(len(FEATURE_COLUMNS),)
        ),
        tf.keras.layers.Dense(
            units,
            activation="relu"
        ),
        tf.keras.layers.Dense(
            max(8, units // 2),
            activation="relu"
        ),
        tf.keras.layers.Dense(1)
    ])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        ),
        loss="mse",
        metrics=["mae"]
    )

    return model

nn_population = [
    np.random.uniform(0, 1, 2)
    for _ in range(6)
]

nn_best_fitness = float("inf")
nn_best_antibody = None
nn_fitness_history = []

def nn_fitness(antibody):

    units, learning_rate = decode_nn_antibody(
        antibody
    )

    tf.keras.backend.clear_session()
    tf.random.set_seed(SEED)

    model = build_nn(
        units,
        learning_rate
    )

    early_stop = tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    )

    model.fit(
        X_train_nn,
        y_train.to_numpy(),
        validation_data=(
            X_val_nn,
            y_val.to_numpy()
        ),
        epochs=60,
        batch_size=8,
        callbacks=[early_stop],
        verbose=0,
        shuffle=False
    )

    pred = np.maximum(
        model.predict(
            X_val_nn,
            verbose=0
        ).flatten(),
        0
    )

    return float(
        np.sqrt(
            mean_squared_error(y_val, pred)
        )
    )

print("\nStarting AIS Neural Network Optimization...")

for generation in range(5):

    scored = [
        (
            nn_fitness(antibody),
            antibody.copy()
        )
        for antibody in nn_population
    ]

    scored.sort(
        key=lambda item: item[0]
    )

    if scored[0][0] < nn_best_fitness:

        nn_best_fitness = scored[0][0]
        nn_best_antibody = scored[0][1].copy()

    nn_fitness_history.append(
        nn_best_fitness
    )

    print(
        f"NN Generation {generation + 1}/5 | "
        f"Validation RMSE: {nn_best_fitness:.4f}"
    )

    elites = [
        item[1].copy()
        for item in scored[:2]
    ]

    new_population = elites.copy()

    for elite in elites:

        for _ in range(2):

            clone = np.clip(
                elite + np.random.normal(
                    0, 0.15, size=2
                ),
                0,
                1
            )

            new_population.append(clone)

    nn_population = new_population[:6]

best_units, best_lr = decode_nn_antibody(
    nn_best_antibody
)

print("\nBest NN Units:", best_units)
print("Best NN Learning Rate:", best_lr)

# Fit a fresh scaler on train + validation.
final_nn_imputer = SimpleImputer(
    strategy="median"
)

final_nn_scaler = StandardScaler()

X_tv_nn = final_nn_imputer.fit_transform(
    X_train_val
)

X_tv_nn = final_nn_scaler.fit_transform(
    X_tv_nn
)

X_test_final_nn = final_nn_scaler.transform(
    final_nn_imputer.transform(X_test)
)

tf.keras.backend.clear_session()
tf.random.set_seed(SEED)

ais_nn = build_nn(
    best_units,
    best_lr
)

ais_nn.fit(
    X_tv_nn,
    y_train_val.to_numpy(),
    epochs=80,
    batch_size=8,
    verbose=0,
    shuffle=False
)

ais_nn_pred = np.maximum(
    ais_nn.predict(
        X_test_final_nn,
        verbose=0
    ).flatten(),
    0
)

ais_nn_metrics = evaluate(
    y_test,
    ais_nn_pred
)

print("\nAIS Neural Network Performance:")
print(ais_nn_metrics)

# ==============================================================
# 12. MODEL COMPARISON
# ==============================================================

metrics = {
    "Baseline Random Forest": baseline_metrics,
    "AIS Random Forest": ais_rf_metrics,
    "AIS Neural Network": ais_nn_metrics
}

comparison_df = pd.DataFrame(
    metrics
).T

comparison_df.index.name = "Model"

comparison_df = comparison_df.sort_values(
    "RMSE"
)

comparison_df.to_csv(
    ROOT / "ais_model_comparison.csv"
)

best_model_name = comparison_df.index[0]

print("\nMODEL COMPARISON:")
print(comparison_df)

print("\nBest Model:", best_model_name)

# ==============================================================
# 13. ACCURACY / PERFORMANCE GRAPH
# ==============================================================

fig, axes = plt.subplots(
    1, 3,
    figsize=(18, 5)
)

metric_names = [
    "R2",
    "MAE",
    "RMSE"
]

titles = [
    "R² Score - Higher is Better",
    "MAE - Lower is Better",
    "RMSE - Lower is Better"
]

for ax, metric, title in zip(
    axes,
    metric_names,
    titles
):

    sns.barplot(
        x=comparison_df.index,
        y=comparison_df[metric].values,
        ax=ax,
        color="teal"
    )

    ax.set_title(title)
    ax.set_ylabel(metric)
    ax.set_xlabel("Models")

    ax.tick_params(
        axis="x",
        rotation=25
    )

    for container in ax.containers:
        ax.bar_label(
            container,
            fmt="%.3f",
            padding=3
        )

plt.suptitle(
    "AIS Rainfall Model Performance",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    ROOT / "ais_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 14. COMPARISON GRAPH
# ==============================================================

fig, ax = plt.subplots(
    figsize=(12, 6)
)

x = np.arange(
    len(comparison_df)
)

width = 0.35

bars1 = ax.bar(
    x - width / 2,
    comparison_df["MAE"],
    width,
    label="MAE",
    color="royalblue"
)

bars2 = ax.bar(
    x + width / 2,
    comparison_df["RMSE"],
    width,
    label="RMSE",
    color="darkorange"
)

ax.set_xticks(x)

ax.set_xticklabels(
    comparison_df.index,
    rotation=15
)

ax.set_title(
    "Baseline vs AIS Optimized Models",
    fontsize=16
)

ax.set_ylabel("Prediction Error")

ax.legend()

ax.bar_label(
    bars1,
    fmt="%.2f",
    padding=3
)

ax.bar_label(
    bars2,
    fmt="%.2f",
    padding=3
)

plt.tight_layout()

plt.savefig(
    ROOT / "ais_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 15. CORRELATION HEATMAP
# ==============================================================

heatmap_columns = [
    "Total_rainfall",
    "month",
    "month_sin",
    "month_cos",
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_6",
    "lag_12",
    "rolling_mean_3",
    "rolling_mean_6",
    "rolling_std_3",
    "rolling_std_6"
]

correlation = featured[
    heatmap_columns
].corr()

plt.figure(
    figsize=(13, 10)
)

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    square=True
)

plt.title(
    "AIS Rainfall Feature Correlation Heatmap",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    ROOT / "ais_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 16. AIS FITNESS CONVERGENCE GRAPH
# ==============================================================

plt.figure(
    figsize=(12, 6)
)

plt.plot(
    range(1, len(fitness_history) + 1),
    fitness_history,
    marker="o",
    linewidth=2,
    label="AIS Random Forest"
)

plt.plot(
    range(1, len(nn_fitness_history) + 1),
    nn_fitness_history,
    marker="s",
    linewidth=2,
    label="AIS Neural Network"
)

plt.title(
    "AIS Optimization Fitness Convergence",
    fontsize=16
)

plt.xlabel("Generation")
plt.ylabel("Best Validation RMSE")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    ROOT / "ais_fitness_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 17. RESULT CSV
# ==============================================================

result_df = pd.DataFrame({
    "Month-Year": test_dates.to_numpy(),
    "Actual_Rainfall": y_test.to_numpy(),
    "Baseline_Prediction": baseline_pred,
    "AIS_RF_Prediction": ais_rf_pred,
    "AIS_NN_Prediction": ais_nn_pred
})

prediction_lookup = {
    "Baseline Random Forest": baseline_pred,
    "AIS Random Forest": ais_rf_pred,
    "AIS Neural Network": ais_nn_pred
}

result_df["Best_Model_Prediction"] = (
    prediction_lookup[best_model_name]
)

result_df["Absolute_Error"] = np.abs(
    result_df["Actual_Rainfall"]
    - result_df["Best_Model_Prediction"]
)

result_df["Squared_Error"] = (
    result_df["Actual_Rainfall"]
    - result_df["Best_Model_Prediction"]
) ** 2

result_df["Month-Year"] = (
    result_df["Month-Year"].dt.strftime(
        "%Y-%m"
    )
)

result_df.to_csv(
    ROOT / "ais_result.csv",
    index=False
)

print("\nAIS RESULT CSV:")
print(result_df.head())

# ==============================================================
# 18. RESULT GRAPH
# ==============================================================

plt.figure(
    figsize=(15, 7)
)

plt.plot(
    test_dates,
    y_test,
    label="Actual Rainfall",
    color="black",
    linewidth=2.5,
    marker="o"
)

plt.plot(
    test_dates,
    baseline_pred,
    label="Baseline Random Forest",
    linestyle="--",
    marker="s"
)

plt.plot(
    test_dates,
    ais_rf_pred,
    label="AIS Random Forest",
    linestyle="--",
    marker="^"
)

plt.plot(
    test_dates,
    ais_nn_pred,
    label="AIS Neural Network",
    linestyle=":",
    marker="d"
)

plt.title(
    "AIS Actual vs Predicted Rainfall",
    fontsize=16
)

plt.xlabel("Month-Year")
plt.ylabel("Total Rainfall")

plt.legend()
plt.grid(True, alpha=0.3)

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    ROOT / "ais_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 19. REFIT FINAL MODELS ON ALL DATA
# ==============================================================

# Final forecasting models use all historical data.
# The reported test metrics above remain unchanged.

final_rf = create_rf(best_params)

final_rf.fit(X, y)

all_nn_imputer = SimpleImputer(
    strategy="median"
)

all_nn_scaler = StandardScaler()

X_all_nn = all_nn_imputer.fit_transform(X)
X_all_nn = all_nn_scaler.fit_transform(X_all_nn)

tf.keras.backend.clear_session()
tf.random.set_seed(SEED)

final_nn = build_nn(
    best_units,
    best_lr
)

final_nn.fit(
    X_all_nn,
    y.to_numpy(),
    epochs=80,
    batch_size=8,
    verbose=0,
    shuffle=False
)

# ==============================================================
# 20. SAVE AIS PKL AND H5
# ==============================================================

PKL_PATH = ROOT / "ais_rainfall_model.pkl"

joblib.dump(
    {
        "model": final_rf,
        "model_name": "AIS Random Forest",
        "feature_columns": FEATURE_COLUMNS,
        "best_parameters": best_params,
        "nn_imputer": all_nn_imputer,
        "nn_scaler": all_nn_scaler,
        "selected_best_model": best_model_name
    },
    PKL_PATH
)

H5_PATH = ROOT / "ais_rainfall_model.h5"

final_nn.save(str(H5_PATH))

print("\nSaved:", PKL_PATH)
print("Saved:", H5_PATH)

# ==============================================================
# 21. FUTURE FEATURE CREATION
# ==============================================================

def build_future_features(history, next_date):

    future_row = pd.DataFrame({
        "Month-Year": [next_date],
        "Total_rainfall": [np.nan]
    })

    temp = pd.concat(
        [history, future_row],
        ignore_index=True
    )

    temp_features = create_features(temp)

    return temp_features[
        FEATURE_COLUMNS
    ].iloc[[-1]]

def predict_future(features):

    if best_model_name == "AIS Neural Network":

        transformed = all_nn_imputer.transform(
            features
        )

        transformed = all_nn_scaler.transform(
            transformed
        )

        prediction = final_nn.predict(
            transformed,
            verbose=0
        ).flatten()[0]

    elif best_model_name == "AIS Random Forest":

        prediction = final_rf.predict(
            features
        )[0]

    else:

        # Refit baseline on all historical data.
        prediction = final_baseline.predict(
            features
        )[0]

    return max(
        float(prediction),
        0.0
    )

final_baseline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        RandomForestRegressor(
            n_estimators=100,
            max_depth=8,
            min_samples_leaf=2,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

final_baseline.fit(X, y)

# ==============================================================
# 22. FORECAST NEXT 12 MONTHS
# ==============================================================

forecast_history = df.copy()

future_records = []

FORECAST_MONTHS = CONFIG["forecast_months"]

for step in range(FORECAST_MONTHS):

    next_date = (
        forecast_history[
            "Month-Year"
        ].iloc[-1]
        + pd.DateOffset(months=1)
    )

    future_X = build_future_features(
        forecast_history,
        next_date
    )

    future_prediction = predict_future(
        future_X
    )

    future_records.append({
        "Month-Year": next_date,
        "Predicted_Rainfall": round(
            future_prediction,
            3
        ),
        "Model_Used": best_model_name
    })

    forecast_history = pd.concat([
        forecast_history,
        pd.DataFrame({
            "Month-Year": [next_date],
            "Total_rainfall": [
                future_prediction
            ]
        })
    ], ignore_index=True)

prediction_df = pd.DataFrame(
    future_records
)

prediction_df.to_csv(
    ROOT / "ais_prediction.csv",
    index=False,
    date_format="%Y-%m"
)

print("\nAIS FUTURE RAINFALL PREDICTIONS:")
print(prediction_df)

# ==============================================================
# 23. PREDICTION GRAPH
# ==============================================================

historical_plot = df.tail(36)

plt.figure(
    figsize=(15, 7)
)

plt.plot(
    historical_plot["Month-Year"],
    historical_plot["Total_rainfall"],
    label="Historical Rainfall",
    color="royalblue",
    linewidth=2,
    marker="o"
)

plt.plot(
    prediction_df["Month-Year"],
    prediction_df["Predicted_Rainfall"],
    label="AIS Forecast",
    color="crimson",
    linewidth=2.5,
    marker="s",
    linestyle="--"
)

plt.plot(
    [
        historical_plot["Month-Year"].iloc[-1],
        prediction_df["Month-Year"].iloc[0]
    ],
    [
        historical_plot["Total_rainfall"].iloc[-1],
        prediction_df["Predicted_Rainfall"].iloc[0]
    ],
    color="crimson",
    linestyle="--"
)

plt.axvline(
    historical_plot["Month-Year"].iloc[-1],
    color="gray",
    linestyle=":",
    label="Forecast Start"
)

plt.title(
    "AIS Monthly Rainfall Forecast - Next 12 Months",
    fontsize=16
)

plt.xlabel("Month-Year")
plt.ylabel("Total Rainfall")

plt.legend()
plt.grid(True, alpha=0.3)

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    ROOT / "ais_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# 24. SAVE AIS MODEL INFORMATION
# ==============================================================

model_info = {
    "project": CONFIG["project"],
    "algorithm": CONFIG["algorithm"],
    "created_at": datetime.now().isoformat(),
    "dataset": CSV_PATH.name,
    "total_records": int(len(df)),
    "training_samples": int(len(X_train)),
    "validation_samples": int(len(X_val)),
    "testing_samples": int(len(X_test)),
    "features": FEATURE_COLUMNS,
    "best_rf_parameters": best_params,
    "best_nn_units": best_units,
    "best_nn_learning_rate": best_lr,
    "best_validation_rf_rmse": float(best_fitness),
    "best_validation_nn_rmse": float(nn_best_fitness),
    "selected_best_model": best_model_name,
    "model_metrics": metrics,
    "forecast_months": FORECAST_MONTHS,
    "saved_files": [
        "ais_rainfall_model.pkl",
        "ais_rainfall_model.h5",
        "ais_config.yaml",
        "ais_model_info.json",
        "ais_accuracy_graph.png",
        "ais_comparison_graph.png",
        "ais_heatmap.png",
        "ais_result.csv",
        "ais_result_graph.png",
        "ais_prediction.csv",
        "ais_prediction_graph.png",
        "ais_fitness_graph.png",
        "ais_model_comparison.csv"
    ]
}

with open(
    ROOT / "ais_model_info.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        model_info,
        f,
        indent=4
    )

# ==============================================================
# 25. FINAL SUMMARY
# ==============================================================

print("\n" + "=" * 65)
print("AIS RAINFALL FORECASTING COMPLETED")
print("=" * 65)

print("\nBest Model:", best_model_name)

print("\nModel Performance:")

for name, metric in metrics.items():

    print(
        f"{name:25s} | "
        f"MAE={metric['MAE']:.4f} | "
        f"RMSE={metric['RMSE']:.4f} | "
        f"R2={metric['R2']:.4f}"
    )

print("\nGenerated Files:")

for filename in model_info["saved_files"]:

    path = ROOT / filename

    if path.exists():
        print("[SAVED]", filename)
    else:
        print("[MISSING]", filename)

print("\nOutput Directory:")
print(ROOT)

print("\nALL AIS RESULTS GENERATED SUCCESSFULLY!")
```
